```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras.layers import (
    Dense, 
    LSTM, 
    Dropout, 
    GRU, 
    Embedding, 
    Bidirectional, 
    TimeDistributed, 
    Flatten, 
    SimpleRNN
)
from keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.datasets import imdb
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)
```

```python
x_train.shape, 
```

```python
x_train=pad_sequences(x_train, maxlen=100)
x_test=pad_sequences(x_test, maxlen=100)
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
model = Sequential()
model.add(Embedding(input_dim=10000, output_dim=32, input_length=100))
model.add(SimpleRNN(5, return_sequences=True))
model.add(SimpleRNN(5, return_sequences=True))
model.add(SimpleRNN(5))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
model1 = Sequential()
model1.add(Embedding(input_dim=10000, output_dim=32, input_length=100))
model1.add(LSTM(5, return_sequences=True))
model1.add(LSTM(5, return_sequences=True))
model1.add(LSTM(5))
model1.add(Dense(1, activation='sigmoid'))
model1.summary()
```

```python
model2 = Sequential()
model2.add(Embedding(input_dim=10000, output_dim=32, input_length=100))
model2.add(GRU(5, return_sequences=True))
model2.add(GRU(5, return_sequences=True))
model2.add(GRU(5))
model2.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
history=model.fit(x_train, y_train, epochs=10, batch_size=64, validation_data=(x_test, y_test))
```
