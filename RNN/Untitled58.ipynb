```python
import numpy as np
import pandas as pd
```

```python
docs =[
    "hello india",
    " however",
    "that my means",
    "hello sam",
    "hello nagendra kushwaha",
    "the world fall for me ",
    "it islike i "
]
```

```python

from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer(oov_token="hello_my_firend")
```

```python
tokenizer
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts
```

```python
tokenizer.document_count
```

```python
sequences=tokenizer.texts_to_sequences(docs)
```

```python
sequences
```

```python
from keras.utils import pad_sequences
```

```python
sequences=pad_sequences(sequences, padding='post')
```

```python
sequences
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, Flatten, LSTM,SimpleRNN, GRU
```

```python
model=Sequential()
model.add(Embedding(20, output_dim=2, input_length=5))
model.summary()
```

```python
model.compile(loss='mse', optimizer='adam', metrics=['accuracy'])
```

```python
model.predict(sequences)
```
