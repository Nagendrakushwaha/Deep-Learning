```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
import numpy as np
```

```python
or_data=pd.DataFrame()
and_data=pd.DataFrame()
xor_data=pd.DataFrame()
```

```python
or_data["input"]=[1, 1, 0, 0]
or_data["input2"]=[1, 0, 1, 0]
or_data["output"]=[1, 1, 1, 0]
```

```python
and_data["input"]=[1, 1, 0, 0]
and_data["input2"]=[1, 0, 1, 0]
and_data["output"]=[0, 0, 0, 1]
```

```python
xor_data["input"]=[1, 1, 0, 0]
xor_data["input2"]=[1, 0, 1, 0]
xor_data["output"]=[0, 1, 1, 0]
```

```python
and_data
```

```python
sns.scatterplot(x=and_data["input"], y=and_data["input2"], hue=and_data["output"], s=200)
```

```python
sns.scatterplot(x=xor_data["input"], y=xor_data["input2"], hue=xor_data["output"], s=200)
```

```python
sns.scatterplot(x=or_data["input"], y=or_data["input2"], hue=or_data["output"], s=200)
```

```python
from sklearn.linear_model import Perceptron
clf1=Perceptron()
clf2=Perceptron()
clf3=Perceptron()
clf1.fit(and_data.iloc[:, 0:2].values, and_data.iloc[:, -1].values)
clf2.fit(xor_data.iloc[:, 0:2].values, xor_data.iloc[:, -1].values)
clf3.fit(or_data.iloc[:, 0:2].values, or_data.iloc[:, -1].values)
```

```python
print(clf1.coef_, clf1.intercept_)
print(clf2.coef_, clf2.intercept_)
print(clf3.coef_, clf3.intercept_)
```
